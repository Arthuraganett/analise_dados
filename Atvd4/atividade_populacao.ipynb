# Atividade — População 2010 x 2022
Leitura da tabela, agregação por estado, ordenação pelo crescimento e análise por município.

## 1 a 4 — Abrir o projeto e ler a tabela

```python
import pandas as pd
import os


caminho_arquivo = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

tabela_bruta = pd.read_excel(caminho_arquivo, sheet_name="Municípios", header=None)

base_municipios = tabela_bruta.iloc[3:].copy()
base_municipios.columns = tabela_bruta.iloc[2]
base_municipios = base_municipios.iloc[:, 1:]

base_municipios.columns = [
    "UF",
    "COD_UF",
    "COD_MUNIC",
    "NOME_MUNICIPIO",
    "POP_2010_SINOPSE",
    "POP_2010_COMPATIBILIZADA",
    "POP_2022"
]

base_municipios = base_municipios.dropna(subset=["UF", "NOME_MUNICIPIO"]).copy()

for nome_coluna in ["POP_2010_COMPATIBILIZADA", "POP_2022"]:
    base_municipios[nome_coluna] = pd.to_numeric(base_municipios[nome_coluna], errors="coerce")

base_municipios.head()
```

## 5 e 6 — Agregar por estado e calcular o crescimento

```python


resumo_estados = (
    base_municipios.groupby("UF")[["POP_2010_COMPATIBILIZADA", "POP_2022"]]
    .sum()
    .reset_index()
)


resumo_estados["CRESCIMENTO"] = (
    resumo_estados["POP_2022"]
    - resumo_estados["POP_2010_COMPATIBILIZADA"]
)


resumo_estados = resumo_estados.sort_values(
    "CRESCIMENTO", ascending=False
).reset_index(drop=True)

resumo_estados
```

## 7 — Salvar a tabela por estado em CSV

```python


resumo_estados.to_csv(
    "populacao_por_estado_crescimento.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)
```

## 8 e 9 — Organizar por município e calcular o crescimento

```python


resumo_municipios = base_municipios[
    ["UF", "COD_MUNIC", "NOME_MUNICIPIO",
     "POP_2010_COMPATIBILIZADA", "POP_2022"]
].copy()


resumo_municipios["CRESCIMENTO"] = (
    resumo_municipios["POP_2022"]
    - resumo_municipios["POP_2010_COMPATIBILIZADA"]
)


resumo_municipios = resumo_municipios.sort_values(
    "CRESCIMENTO", ascending=False
).reset_index(drop=True)

resumo_municipios
```

## 10 — Salvar a tabela por município em CSV

```python


resumo_municipios.to_csv(
    "populacao_por_municipio_crescimento.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)
```
